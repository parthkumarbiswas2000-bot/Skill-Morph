# Task 2.2 Create and Fix Missing Value

```python
df_practice = df.copy()
print("Original missing values:", df_practice.isnull().sum())
np.random.seed(42)
missing_math_idx = np.random.choice(df_practice.index, 50, replace=False)
missing_writing_idx = np.random.choice(df_practice.index, 30, replace=False)

df_practice.loc[missing_math_idx, 'math score'] = np.nan
df_practice.loc[missing_writing_idx, 'writing score'] = np.nan

print("Missing values after modification:", df_practice.isnull().sum())
```

```python
#Math Score Null value replacing with median
math_median = df_practice['math score'].median()
df_practice['math score'] = df_practice['math score'].fillna(math_median)

print(df_practice['math score'].isna().sum())
```

```python
#Writing Score Null value replacing with mean

writing_mean = df_practice['writing score'].mean()
df_practice['writing score'] = df_practice['writing score'].fillna(writing_mean)

print(df_practice['writing score'].isna().sum())
```

```python
no_missing_values = df_practice.isnull().sum()
print(no_missing_values)
```

# Task 3.2: Basic Groupby Analysis

```python
# Group By Gender
gender_grouped = df_practice.groupby('gender').mean(numeric_only=True)

print(gender_grouped)

print('\n')

# Second process for average of all subject
average_all_subject = df_practice.mean(numeric_only=True)
print(average_all_subject)

print('\n')
# Count of students
count_students = df_practice['gender'].value_counts()
print(count_students)
```

```python
# Group By Test Prepartion Course
test_prep_grouped = df_practice.groupby('test preparation course')

mean_each_subject = test_prep_grouped.mean(numeric_only=True)
print(mean_each_subject)

score_improvement = mean_each_subject.loc['completed'] - mean_each_subject.loc['none']

most_improved_subject = score_improvement.idxmax()
max_gain = score_improvement.max()

print("Average score improvement by subject:")
print(score_improvement)
print(f"\n📈 Subject that benefited most from test prep: {most_improved_subject} (+{max_gain:.2f} points)")
```

```python
# Group By Parental Education

parental_level_of_education_grouped = df_practice.groupby('parental level of education')

mean_math_subject = parental_level_of_education_grouped['math score'].mean()
print("Math Score based Average", mean_math_subject)
```

```python
rank_education_level_math_performance = mean_math_subject.sort_values(ascending=False)
print(rank_education_level_math_performance)
```

```python
# Group by 'lunch' type and show:

lunch_grouped = df_practice.groupby('lunch')

average_score_each_subject = lunch_grouped.mean(numeric_only=True)
print(average_score_each_subject)

print('---------')

# Count of Students in lunch
count_lunch_type = df_practice['lunch'].value_counts()
print(count_lunch_type)
```
