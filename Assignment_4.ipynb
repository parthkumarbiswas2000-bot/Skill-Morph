# Task 1 Matplotlib

## **Bar_Chart**

```python

# Abonoment Type
Abonoment_count = gym_members['abonoment_type'].value_counts()

# Simple bar chart
plt.bar(['Premium', 'Standard'], Abonoment_count.values)
plt.title('Abonoment Type')
plt.savefig("/content/drive/MyDrive/Datasets_SkillMorph/abonoment_type.jpg", dpi=300, bbox_inches='tight')
plt.show()
```

```python
# With color
plt.bar(['Premium', 'Standard'], Abonoment_count.values, color=['magenta', 'blue'])
plt.ylabel('Count')
plt.title('Abonoment Type')
plt.show()
```

```python
# Group by gender, average math score
avg_gym_time = gym_members.groupby('abonoment_type')['avg_time_in_gym'].mean()

# Bar plot
plt.grid(True, which="both", linestyle=":", linewidth=0.7, alpha=0.7)
plt.bar(avg_gym_time.index, avg_gym_time.values)
plt.title('Average Gym Time by Abonoment Type')
plt.xlabel('abonoment_type')
plt.ylabel('Average Gym Time')
plt.show()
```

## **Histogram**

```python
# Histogram
plt.hist(gym_members['avg_time_in_gym'])
plt.title('Average Gym Time Distribution')
plt.show()
```

```python
# With more bins
plt.hist(gym_members['avg_time_in_gym'], bins=20, color='green')
plt.xlabel('Time')
plt.ylabel('Count')
plt.title('Average Time In Gym')
plt.show()
```

## **Scatter Plot**

```python
# Age vs visits per week
plt.scatter(gym_members['Age'], gym_members['visit_per_week'])
plt.grid(True)
plt.xlabel('Age')
plt.ylabel('Visits per Week')
plt.title('Age vs Visits per Week')
plt.show()
#plt.savefig("/content/drive/MyDrive/Datasets_SkillMorph/age_vs_visit_per_week.jpg", dpi=300, bbox_inches='tight
```

```python
sample = gym_members.head(100)
plt.scatter(sample['Age'], sample['visit_per_week'],alpha=0.7)
plt.xlabel('Age')
plt.ylabel('Visits per Week')
plt.title('Age vs Visits per Week')
plt.show()
```

## **Pie Chart**

```python
attend_group = gym_members['attend_group_lesson'].value_counts()

plt.pie(attend_group.values, labels=['True', 'False'])
plt.title('attend_group_lesson')
plt.show()
```

```python
plt.pie(attend_group.values, labels=['True', 'False'],autopct='%1.1f%%')
plt.title('attend_group_lesson')
plt.show()
```

```python
name_personal_trainer_counts = gym_members['name_personal_trainer'].value_counts()
plt.pie(name_personal_trainer_counts, labels=name_personal_trainer_counts.index, autopct='%1.2f%%')
plt.axis('equal')  # Makes the pie chart a perfect circle
plt.tight_layout()
plt.title('fav_group_lesson')
plt.show()
```

## **Multiple Graph**

```python
# Create 2x2 grid
fig, axes = plt.subplots(2, 2, figsize=(10, 8))

# Plot 1: Histogram
axes[0, 0].hist(gym_members['avg_time_in_gym'], bins=15)
axes[0, 0].set_title('Average Time in Gym')

# Plot 2: Bar chart
gym_members['gender'].value_counts().plot(kind='bar', ax=axes[0, 1])
axes[0, 1].set_title('Gender')

# Plot 3: Scatter
axes[1, 0].scatter(gym_members['Age'][:50], gym_members['avg_time_in_gym'][:50])
axes[1, 0].grid(True)
axes[1, 0].set_xlabel('Age')
axes[1, 0].set_ylabel('Average Time in Gym')
axes[1, 0].set_title('Age vs Average Time in gym')


# Plot 4: Line
axes[1, 1].plot(gym_members['days_per_week'][:30])
axes[1, 1].set_title('days_per_week')

plt.tight_layout()
plt.show()

```

## **Count Plot**

```python
plt.figure(figsize=(21, 7)) # Increase figure size
sns.countplot(data=gym_members, x='days_per_week')
plt.grid(True, which="major", linestyle="solid", linewidth=0.7, alpha=1.0)
plt.xticks(rotation=90)
plt.tight_layout()
plt.title('Days_per_week Count')
plt.savefig("/content/drive/MyDrive/Datasets_SkillMorph/count_plot.jpg", dpi=300, bbox_inches='tight')
plt.show()
'''
# Display the saved image in the notebook to confirm
from IPython.display import Image
display(Image(filename="/content/drive/MyDrive/Datasets_SkillMorph/count_plot.jpg"))

'''
```

```python
gym_members['days_per_week_clean'] = gym_members['days_per_week'].apply(
    lambda x: len(str(x).split(',')) if pd.notnull(x) else 0
)

# Now plot the cleaned version
sns.countplot(data=gym_members, x='days_per_week_clean')
plt.grid(True, which="major", linestyle="--", linewidth=0.7, alpha=0.7)
plt.title('Cleaned Days per Week Count')
plt.xlabel('Days per Week')
plt.show()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

day_order = [1, 2, 3, 4, 5, 6, 7]

# Plot the cleaned version
plt.figure(figsize=(9, 5))  # Set appropriate figure size
sns.set_style("whitegrid")  # Cleaner background

# Create count plot
sns.countplot(data=gym_members, x='days_per_week_clean', palette='Blues_d', order=day_order)

# Set title and labels
plt.title('Cleaned Days per Week Count')
plt.xlabel('Days per Week')

# Add grid only on Y-axis (more readable)
plt.grid(True, axis='y',linestyle="--", linewidth=0.5, alpha=0.7)

# Titles and labels
plt.title('Gym Attendance Frequency (Days per Week)', fontsize=14, weight='bold')
plt.xlabel('Number of Days per Week', fontsize=12)
plt.ylabel('Number of Members', fontsize=12)

# Optional: Show exact counts on top of each bar
for container in plt.gca().containers:
    plt.bar_label(container, fmt='%d', label_type='edge', fontsize=9)

# Final layout tweaks
plt.tight_layout()
plt.show()
```
